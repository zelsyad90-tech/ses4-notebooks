### Worked example: total, average and maximum

```python
total = 0
count = 0
largest = 0

while True:
    text = input("Number (or done): ").strip()
    if text == "done":
        break
    number = float(text)
    total += number
    count += 1
    if count == 1 or number > largest:
        largest = number

if count > 0:
    print("Count:", count)
    print(f"Average: {total / count:.2f}")
    print("Max:", largest)
```

### Worked example: password strength checker

```python
password = input("Password: ")

digits = 0
capitals = 0
for ch in password:
    if ch.isdigit():
        digits += 1
    elif ch.isupper():
        capitals += 1

if len(password) >= 8 and digits > 0 and capitals > 0:
    print("Strong")
elif len(password) >= 8:
    print("Medium: add a digit and a capital letter")
else:
    print("Weak: use at least 8 characters")
```

### Early return and guard clauses

### Scope: where names live

### The call stack

```python
def apply_tax(amount):
    return amount * 1.25

def total_price(price, qty):
    subtotal = price * qty
    return apply_tax(subtotal)

print(total_price(50, 2))
```

### Docstrings and type hints

```python
def calculate_tip(bill: float, percent: int = 15) -> float:
    """Return the tip for a bill.

    percent is a percentage, e.g. 15 for 15%.
    """
    return bill * percent / 100

print(calculate_tip(250))
help(calculate_tip)
```

### Pure functions and side effects

```python
# Pure: the result depends only on the inputs
def share(total, people):
    return total / people

# Side effects: prints, and changes a global
paid = 0
def pay(amount):
    global paid
    paid += amount
    print(f"Paid {amount}")

pay(share(300, 3))
print(paid)
```

### Testing functions with assert

```python
def is_leap(year):
    if year % 400 == 0:
        return True
    if year % 100 == 0:
        return False
    return year % 4 == 0

assert is_leap(2024)
assert not is_leap(2026)
assert not is_leap(1900)
assert is_leap(2000)
print("All checks passed")

# A wrong expectation stops the cell:
assert is_leap(2100), "2100 is not a leap year"
```

### Worked example: the tip calculator, refactored

```python
def calculate_tip(bill, percent):
    return bill * percent / 100

def split_bill(total, people):
    return total / people

def main():
    bill = float(input("Bill amount: "))
    percent = int(input("Tip %: "))
    people = int(input("People: "))
    tip = calculate_tip(bill, percent)
    each = split_bill(bill + tip, people)
    print(f"Tip: {tip:.2f} | Each pays: {each:.2f}")

main()
```

### Build-along 1: the code

```python
import random

secret = random.randint(1, 100)
tries = 0
guess = 0

while guess != secret:
    guess = int(input("Your guess (1-100): "))
    tries += 1
    if guess < secret:
        print("Too low")
    elif guess > secret:
        print("Too high")

print(f"Correct! {secret} in {tries} tries.")
```
