### Worked example: word frequency

```python
text = """The cat sat on the mat. The dog sat on the log.
The cat saw the dog, and the dog saw the cat."""

clean = text.lower().replace(".", "").replace(",", "")
words = clean.split()

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

top = sorted(counts, key=counts.get, reverse=True)[:5]
for word in top:
    print(f"{word:<5} {counts[word]}")
```

### Worked example: save and load a to-do list

```python
import json
from pathlib import Path

def save_tasks(tasks, path):
    text = json.dumps(tasks, indent=2)
    Path(path).write_text(text, encoding="utf-8")

def load_tasks(path):
    text = Path(path).read_text(encoding="utf-8")
    return json.loads(text)
```

```python
tasks = [
    {"title": "buy milk", "done": False},
    {"title": "finish lab", "done": True},
]
save_tasks(tasks, "todo.json")

for task in load_tasks("todo.json"):
    if task["done"]:
        print("[x]", task["title"])
    else:
        print("[ ]", task["title"])
```

## Exceptions

### try and except

```python
try:
    number = int("abc")
except ValueError:
    print("Not a whole number")

try:
    result = 10 / 0
except ZeroDivisionError as error:
    print("Could not divide:", error)

prices = {"tea": 12}
try:
    print(prices["coffee"])
except KeyError:
    print("No price for coffee")

print("The program carries on")
```

### else and finally

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("cannot divide by zero")
        return None
    else:
        return result
    finally:
        print("done trying")

print(safe_divide(10, 4))
print(safe_divide(1, 0))
```

### raise: signalling your own errors

```python
def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError("amount must be positive")
    if amount > balance:
        raise ValueError(f"only {balance} available")
    return balance - amount

print(withdraw(100, 30))

try:
    withdraw(100, 250)
except ValueError as error:
    print("Refused:", error)

withdraw(100, -5)
```

### Worked example: a safe JSON loader

```python
import json
from pathlib import Path

def load_json(path, default):
    try:
        text = Path(path).read_text(encoding="utf-8")
        return json.loads(text)
    except FileNotFoundError:
        print(f"{path}: not found, using default")
        return default
    except json.JSONDecodeError as error:
        print(f"{path}: broken JSON ({error.msg})")
        return default
```

### The safe loader on three files

```python
Path("good.json").write_text('{"theme": "dark"}', encoding="utf-8")
Path("bad.json").write_text('{"theme": dark}', encoding="utf-8")

print(load_json("good.json", {}))
print(load_json("bad.json", {}))
print(load_json("missing.json", {}))
```

### self, attributes and methods

```python
class Student:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def average(self):
        return sum(self.scores) / len(self.scores)

mona = Student("Mona", [90, 85])
omar = Student("Omar", [70, 78])
omar.scores.append(95)

print(mona.name, mona.average())
print(omar.name, omar.average())
```

### Worked example: a bank account

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError(f"only {self.balance} available")
        self.balance -= amount

    def __str__(self):
        return f"{self.owner}: {self.balance}"
```

```python
account = BankAccount("Mona", 100)
account.deposit(50)
account.withdraw(30)
print(account)

try:
    account.withdraw(500)
except ValueError as error:
    print("Refused:", error)
```

### Dataclasses: less typing for data

```python
from dataclasses import dataclass

@dataclass
class Task:
    title: str
    done: bool = False

a = Task("buy milk")
b = Task("finish lab", done=True)
print(a)
print(b.title, b.done)
print(a == Task("buy milk"))
```

### Inheritance: extend and override

```python
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, rate=0.05):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        self.deposit(self.balance * self.rate)

    def __str__(self):
        return f"{self.owner} (savings): {self.balance:.2f}"

savings = SavingsAccount("Omar", 200)
savings.add_interest()
print(savings)
print(isinstance(savings, BankAccount))
```

### Worked example: a to-do list class

```python
import json
from dataclasses import dataclass, asdict

@dataclass
class Task:
    title: str
    done: bool = False

class TodoList:
    def __init__(self):
        self.tasks = []
    def add(self, title):
        self.tasks.append(Task(title))
    def complete(self, title):
        for task in self.tasks:
            if task.title == title:
                task.done = True
    def save(self, path):
        data = [asdict(task) for task in self.tasks]
        with open(path, "w", encoding="utf-8") as f:
            json.dump(data, f, indent=2)
```

```python
todo = TodoList()
todo.add("buy milk")
todo.add("finish lab")
todo.complete("finish lab")
todo.save("todo.json")

with open("todo.json", encoding="utf-8") as f:
    print(f.read())
```

### Worked example: the to-do list as a module

```python
%%writefile todo.py
from dataclasses import dataclass

@dataclass
class Task:
    title: str
    done: bool = False

class TodoList:
    def __init__(self):
        self.tasks = []
    def add(self, title):
        self.tasks.append(Task(title))
    def complete(self, title):
        for task in self.tasks:
            if task.title == title:
                task.done = True
    def pending(self):
        return [t.title for t in self.tasks if not t.done]
```

```python
from todo import TodoList

todo = TodoList()
todo.add("buy milk")
todo.add("finish lab")
todo.complete("buy milk")
print(todo.pending())
```
